```python
import pandas as pd

df = pd.read_csv("exercise.csv")

df.head()
```

```python
df.info()
```

```python
df.shape
```

```python
df = df.drop("Unnamed: 0", axis=1)

df.head()
```

```python
df.duplicated().sum()
```

```python
import matplotlib.pyplot as plt


avg_pulse = df.groupby("diet")["pulse"].mean()


avg_pulse.plot(kind="bar")

plt.title("Average Pulse by Diet Type")
plt.ylabel("Average Pulse")
plt.show()
```

```python
avg_time = df.groupby("time")["pulse"].mean()

avg_time.plot(kind="bar")

plt.title("Average Pulse by Time")
plt.ylabel("Average Pulse")
plt.show()
```
