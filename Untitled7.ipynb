```python
import pandas as pd

df = pd.read_csv('Student_performance_data _.csv')

df.head()
```

```python
print("Shape:", df.shape)

print("\nColumns:")
print(df.columns)

print("\nData Types:")
print(df.dtypes)

print("\nMissing Values:")
print(df.isnull().sum())

print("\nDuplicate Rows:", df.duplicated().sum())
```

```python
df.describe()
```

```python
import numpy as np

# introduce missing values in 5% of the data
df.loc[df.sample(frac=0.05).index, 'StudyTimeWeekly'] = np.nan
```

```python
df.isnull().sum()
```

```python
df['StudyTimeWeekly'].fillna(df['StudyTimeWeekly'].mean(), inplace=True)
```

```python
df.isnull().sum()
```

```python
num_cols = ['Age', 'StudyTimeWeekly', 'Absences', 'GPA']
```

```python
for col in num_cols:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    outliers = df[(df[col] < lower) | (df[col] > upper)]

    print(f"{col} outliers count:", len(outliers))
```

```python
for col in num_cols:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    df = df[(df[col] >= lower) & (df[col] <= upper)]
```

```python
df.shape
```

```python
import numpy as np

# add artificial outliers
df.loc[df.sample(frac=0.02).index, 'Absences'] = df['Absences'].max() * 3
```

```python
df['Absences'].describe()
```

```python
for col in ['Age', 'StudyTimeWeekly', 'Absences', 'GPA']:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    df = df[(df[col] >= lower) & (df[col] <= upper)]
```

```python
df.shape
```

```python
from sklearn.preprocessing import MinMaxScaler

scaler = MinMaxScaler()

df[['Age', 'StudyTimeWeekly', 'Absences', 'GPA']] = scaler.fit_transform(
    df[['Age', 'StudyTimeWeekly', 'Absences', 'GPA']]
)

df.head()
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

df[['Age', 'StudyTimeWeekly', 'Absences', 'GPA']] = scaler.fit_transform(
    df[['Age', 'StudyTimeWeekly', 'Absences', 'GPA']]
)

df.head()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

plt.figure(figsize=(8,6))
sns.heatmap(df[['Age', 'StudyTimeWeekly', 'Absences', 'GPA']].corr(), annot=True, cmap='coolwarm')
plt.show()
```

```python
from sklearn.decomposition import PCA

X = df[['Age', 'StudyTimeWeekly', 'Absences', 'GPA']]

pca = PCA(n_components=2)
principal_components = pca.fit_transform(X)

pca_df = pd.DataFrame(data=principal_components, columns=['PC1', 'PC2'])

pca_df.head()
```

```python
print("Explained variance ratio:", pca.explained_variance_ratio_)
```
